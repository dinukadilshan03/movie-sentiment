# Descriptive Analysis of the Corrected Final Dataset

This notebook summarizes the retained descriptive variables after removing the dropped interaction term from the accepted workflow.

```python
from pathlib import Path
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

sns.set_theme(style='whitegrid', context='notebook')
pd.set_option('display.max_columns', None)
pd.set_option('display.float_format', lambda x: f'{x:,.4f}')

DATA_PATH = Path.cwd().parent / 'data' / 'final' / 'final.corrected.csv'
df = pd.read_csv(DATA_PATH)

KEY_VARS = [
    'log_box_office',
    'audienceScore',
    'initial_combined_sentiment_score',
    'log_initial_review_count',
]
for col in KEY_VARS + ['box_office_num']:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors='coerce')

print(f'Data path: {DATA_PATH}')
print(f'Dataset shape: {df.shape[0]:,} rows x {df.shape[1]:,} columns')
```

## Summary Statistics

```python
summary = df[KEY_VARS].agg(['count', 'mean', 'median', 'std', 'min', 'max']).T.round(4)
summary
```

## Correlation Matrix

```python
corr = df[KEY_VARS].corr(numeric_only=True).round(4)
corr
```

```python
plt.figure(figsize=(8, 6))
sns.heatmap(corr, annot=True, fmt='.4f', cmap='RdBu_r', center=0, square=True)
plt.title('Correlation matrix of retained descriptive variables')
plt.tight_layout()
plt.show()
```

## Univariate and Bivariate Plots

```python
fig, axes = plt.subplots(2, 2, figsize=(12, 8))
for ax, col in zip(axes.flat, KEY_VARS):
    sns.histplot(df[col].dropna(), kde=True, ax=ax)
    ax.set_title(col)
plt.tight_layout()
plt.show()

fig, axes = plt.subplots(1, 3, figsize=(15, 4))
for ax, col in zip(axes, KEY_VARS[1:]):
    sns.regplot(data=df[[col, 'log_box_office']].dropna(), x=col, y='log_box_office', scatter_kws={'alpha': 0.25, 's': 20}, line_kws={'color': 'red'}, ax=ax)
    ax.set_title(f'{col} vs log_box_office')
plt.tight_layout()
plt.show()
```
